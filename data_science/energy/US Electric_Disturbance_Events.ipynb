US Electric Grid Outages
Electric outage incidents in the US power grid from January 2002 to July 2023, including details related to the event start and end time, location, alert criteria, demand loss, and estimated number of people affected.

NOTE: This dataset was compiled exactly as reported by the US Department of Energy. We have taken no measures to clean or prepare it, as our goal is to expose users to the types of data quality issues often encountered in the real world.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'C:\Users\DELL\Desktop\Data\datasets\raws\US Electric Grid new.csv')
df['Area Affected'] = df['Area Affected'].str.strip(';')
```

As Demand Loss is very useful, it doesn't make sense to keep Unknown values. Same thing for Number of Customers Affected.

```python
df = df.drop(df.loc[df['Demand Loss (MW)']=='Unknown'].index)
df = df.drop(df.loc[df['Number of Customers Affected']=='Unknown'].index)
df = df.drop(columns=['Event Year'])
```

Correcting wrong data types for some columns.

```python
df['Number of Customers Affected'] = pd.to_numeric(df['Number of Customers Affected'])
df['Demand Loss (MW)'] = pd.to_numeric(df['Demand Loss (MW)'])
df['Date Event Began'] = pd.to_datetime(df['Date Event Began'])
```

```python
df.reset_index(drop=True,inplace=True)
```
