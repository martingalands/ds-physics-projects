**Context**

As global energy demands rise and climate-driven extreme weather events become more frequent, understanding the resilience of our electrical infrastructure is critical. This dataset tracks major power outages and grid reliability incidents from 2010 to the present, capturing the moments when the lights went out and why.

Content
This dataset includes detailed logs of significant grid disturbances. Key features typically found in this data include:

Event Date/Time: When the disturbance began and ended.
Region/NERC Region: Geographic area or grid authority affected.
Cause: The primary reason for the outage (e.g., Severe Weather, Equipment Failure, Cyber Attack, System Operations).
Impact: Estimated number of customers affected and the total demand loss (Megawatts).
Recovery: Insights into the restoration time for various event types.
Inspiration & Potential Analysis
I created this dataset to provide a clean, long-term window into how our energy systems are evolving. Researchers can use this data to:

Trend Analysis: Are certain types of outages (like weather-related) becoming more frequent?\
Risk Modeling: Build predictive models to estimate the duration of an outage based on its cause.\
Policy Evaluation: Analyze how grid modernization efforts have impacted recovery times over the last decade.\

**Sources**

This data is compiled and cleaned from public energy reliability reports, utility filings, and government infrastructure monitoring archives (such as DOE/EIA records).
